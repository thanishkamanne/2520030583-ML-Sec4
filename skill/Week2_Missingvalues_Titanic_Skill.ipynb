```python
import pandas as pd

# Load dataset
df = pd.read_csv("/content/Titanic-Dataset.csv")

# Fill missing numerical values with column mean
#inplace=True tells pandas to modify the original DataFrame directly, instead of creating a new DataFrame.
df.fillna(df.mean(numeric_only=True), inplace=True)

# Save the updated dataset to a new CSV file
#index=False tells pandas not to save the row numbers (index) into the CSV file.
df.to_csv("placementfilled.csv", index=False)

# Check if any missing values remain
print(df.isnull().sum())

print("\nNew dataset saved as 'plzcementfilled.csv'")
```

```python
import pandas as pd

df = pd.read_csv("/content/Titanic-Dataset.csv")

# Fill missing categorical values with mode
for col in df.select_dtypes(include='object').columns:
    df[col].fillna(df[col].mode()[0], inplace=True)
df.to_csv("placementfilledcat.csv", index=False)
print(df.isnull().sum())
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("/content/placementfilledcat.csv")

# Identify categorical columns
cat_cols = df.select_dtypes(include=['object']).columns

# Apply One-Hot Encoding
df = pd.get_dummies(df, columns=cat_cols, drop_first=True, dtype=int)

# Display first 5 rows
print(df.head())
#print(df)
# Check the new column names
print(df.columns)
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Load dataset
df = pd.read_csv("/content/placementfilledcat.csv")

# Select only two numerical features
X = df[['PassengerId', 'Survived']]

# Apply Standard Scaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert to DataFrame
scaled_df = pd.DataFrame(X_scaled, columns=['PassengerId', 'Survived'])

print("Original Data")
print(X.head())

print("\nAfter Standard Scaling")
print(scaled_df.head())
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Load dataset
df = pd.read_csv("/content/placementfilledcat.csv")

# Select only two numerical features
X = df[['PassengerId', 'Survived']]

# Apply Min-Max Scaler
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# Convert to DataFrame
scaled_df = pd.DataFrame(X_scaled, columns=['PassengerId', 'Survived'])

print("Original Data")
print(X.head())

print("\nAfter Min-Max Scaling")
print(scaled_df.head())
```
